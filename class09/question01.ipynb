#### (A) たまたま発生する相関
- **日本で少子化が進むと、地球温暖化が進む**
  - 日本の少子化によって地球温暖化が引き起こされているわけではない。むしろ、人口が減少して地球温暖化の原因と考えられる二酸化炭素排出量が減り、地球温暖化が進行しないはずである。
- **朝の占いが良いとテストの成績が良い**
  - テストの結果は自分の能力次第であり、運勢とは関係ない。

#### (B) 共通の原因を持つ相関
- **血圧と年収**
  - 年齢という交絡因子が存在するため。
- **一人当たりの年間チョコレート消費量とノーベル賞受賞者数**
  - 一人当たりのGDPという交絡因子が存在するため。

#### (C) 因果関係から発生する相関
- **猫が増えるとネズミが減る**
  - ネズミの天敵である猫の増加によりネズミが捕食されるため。
- **平均気温が上昇すると積雪量が減少する**
  - 気温が低いと積雪量が多くなるため。
