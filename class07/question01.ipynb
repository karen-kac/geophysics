#### **（1）複素数の冪乗**
オイラーの公式 $ e^{ix} $ を利用して、複素数 $ z $ のべきについて複素平面上で考える。

$ z = re^{ix} $ とかく。（$ r $ は正の実数）


\begin{aligned}
\frac{d}{dx}z &= ire^{ix} = iz \\
\frac{d^2}{dx^2}z &= i^2z = -z \\
\frac{d^3}{dx^3}z &= -iz \\
\frac{d^4}{dx^4}z &= -i(iz) = z
\end{aligned}


これらの結果から、一階微分では $ z $ が $ i $ 倍され、二回微分では、 $ -1 $ 倍、三階微分では $ -i $ 倍、四階微分では $ z $ に戻る。

$ z = a + bi $ ($ a, b $ は実数) を複素平面上のベクトル $ (a, b) $ と見て、実軸の正を右方向、虚軸の正を上方向に取った複素平面上で考える。

その一階微分は $ z $ が $ i $ 倍される。これは、反時計周りの $ 90° $ 回転に相当するため、 $ z $ の二回微分は反時計回りの$ 180° $ 回転に相当する。
虚軸が下向きだと時計回りになる。



#### **$(2)$ $\sin(z) = 2$**

オイラーの公式 $e^{i\theta} = \cos\theta + i\sin\theta$ を用いて $\cos\theta$ と $\sin\theta$ を表す。

$$
e^{i\theta} = \cos\theta + i\sin\theta 
$$

$$
e^{-i\theta} = \cos\theta - i\sin\theta \quad 
$$
より
$$
\cos\theta = \frac{e^{i\theta} + e^{-i\theta}}{2} ,\quad \sin\theta = \frac{e^{i\theta} - e^{-i\theta}}{2i} \quad -①
$$

$z = x + iy \quad$ ($x, y$ は実数) として、$y \neq 0 $ とする。


$$
\sin(z) = \sin(x + iy) 
$$

$$
= \sin x \cos(iy) + \cos x \sin(iy) \quad -②
$$
②に①を代入して

$$
= \sin x \frac{e^{i(iy)} + e^{-i(iy)}}{2} + \cos x \frac{e^{i(iy)} - e^{i(iy)}}{2i}\quad 
$$
$$
= 2
$$

すなわち、

$$
\sin x (e^{-y} + e^{y}) - i \cos x (e^{-y} - e^{y}) = 4 -③
$$

について考える。

( $ I $ ) $\cos x = 0$ のとき

$$
x = \frac{\pi}{2} + n\pi \quad (n は整数)
$$

( $ II $ ) $e^y + e^{-y} = 0$ のとき

$$
e^{2y} = 1 \quad \therefore y = 0
$$

$y \neq 0 $
よって$ \sin z = 2 $にはならない

( $ I $ ), ( $ II $ ) より $x = \frac{\pi}{2} + n\pi$

これを③に代入して

$$
\sin\left( \frac{\pi}{2} + n\pi \right) \left( e^{-y} + e^y \right) = 4
$$
両辺に$ e^{y} $をかけて
$$
(-1)^n \left( e^{2y} + 1 \right) = 4 e^{y}
$$

$$
e^{2y} - (-1)^n 4 e^{y} + 1 = 0
$$

解の公式より

$$

$e^{y} > 0 $ より n は偶数であるから

### **オイラー法と改良オイラー法**

#### **オイラー法 (Euler Method)**

オイラー法は、常微分方程式 (ODE) の初期値問題を数値的に解くための最も基本的な方法の一つです。簡単で実装しやすいですが、精度が低く、大きなステップサイズを使用すると誤差が蓄積しやすいという欠点があります。

##### **基本概念**

オイラー法は、微分方程式の解を直線的に近似することに基づいています。具体的には、次のように時間ステップを進めます。

##### **数式**

与えられた微分方程式:
$$ \frac{dy}{dx} = f(x, y) $$

初期条件:
$$ y(x_0) = y_0 $$

オイラー法の更新式:
$$ y_{n+1} = y_n + h f(x_n, y_n) $$

ここで、$h$ はステップサイズです。

#### **改良オイラー法 (Improved Euler Method)**

改良オイラー法（別名: 2次ルンゲクッタ法）は、オイラー法の精度を改善するための手法です。基本的なアイデアは、ステップの両端での傾きを平均することにより、直線近似の精度を向上させることです。

##### **基本概念**

改良オイラー法では、各ステップで傾きを2回評価し、その平均を使用して次のステップを計算します。

##### **数式**

与えられた微分方程式:
$$ \frac{dy}{dx} = f(x, y) $$

初期条件:
$$ y(x_0) = y_0 $$

改良オイラー法の更新式:

1. 中間傾きの計算:
   $$ k1 = f(x_n, y_n) $$
   $$ k2 = f(x_n + h, y_n + h k1) $$

2. 次のステップの計算:
   $$ y_{n+1} = y_n + \frac{h}{2} (k1 + k2) $$

ここで、$h$ はステップサイズです。

#### **比較**

オイラー法は単純で計算が速い一方で、誤差が大きくなりがちです。改良オイラー法は計算コストが少し高くなりますが、精度が大幅に向上します。したがって、精度が求められる場合には改良オイラー法が一般的に選ばれます。


```python
"""
0 ≦ x ≦ 30 の範囲で、解析的に解いた結果（普通に解いた結果）と、
オイラー法で数値的に解いた結果を比較する図
"""
import numpy as np
import matplotlib.pyplot as plt

def analytical_solution(x):
    return np.sin(x)

def euler_method():
    h = 0.1
    x = np.arange(0, 30+h, h)
    y = np.zeros_like(x)
    y[0] = 0

    for i in range(1, len(x)):
        y[i] = y[i-1] + h * np.cos(x[i-1])

    return x, y

# 解析的な解
x_analytical = np.linspace(0, 30, 100)
y_analytical = analytical_solution(x_analytical)

# オイラー法による数値解
x_numerical, y_numerical = euler_method()

# グラフの作成
plt.figure(figsize=(12, 8))
plt.plot(x_analytical, y_analytical, color='deeppink', label='Analytical Solution', linewidth=2)
plt.plot(x_numerical, y_numerical, 'ro-', color='orange', label='Euler Method', markersize=4, linewidth=1)
plt.xlabel('x', fontsize=14)
plt.ylabel('y', fontsize=14)
plt.title('Comparison of Analytical and Numerical Solutions', fontsize=16)
plt.legend(loc='best', fontsize=12)
plt.grid(True)
plt.show()
```

```python
"""
0 ≦ x ≦ 30 の範囲で、解析的に解いた結果（普通に解いた結果）と、
改良オイラー法で数値的に解いた結果を比較する図
"""
def analytical_solution(x):
    return np.sin(x)

def improved_euler_method():
    h = 0.1
    x = np.arange(0, 30+h, h)
    y = np.zeros_like(x)
    y[0] = 0

    for i in range(1, len(x)):
        k1 = np.cos(x[i-1])
        k2 = np.cos(x[i])
        y[i] = y[i-1] + h/2 * (k1 + k2)

    return x, y

# 解析的な解
x_analytical = np.linspace(0, 30, 100)
y_analytical = analytical_solution(x_analytical)

# 改良オイラー法による数値解
x_numerical, y_numerical = improved_euler_method()

# グラフの作成
plt.figure(figsize=(12, 8))
plt.plot(x_analytical, y_analytical, color='deeppink', label='Analytical Solution', linewidth=4)
plt.plot(x_numerical, y_numerical, 'r.-', color='orange', label='Improved Euler Method', markersize=6, linewidth=1)

plt.xlabel('x', fontsize=16)
plt.ylabel('y', fontsize=16)
plt.title('Comparison of Analytical and Numerical Solutions', fontsize=18)
plt.legend(fontsize=14, loc='upper right')
plt.grid(True)
plt.tight_layout()  # レイアウト調整
plt.show()
```
